```python
dominant_age_midpoint = {
    "before_1900": 1890,
    "1901_1910": 1905,
    "1911_1920": 1915,
    "1921_1930": 1925,
    "1931_1940": 1935,
    "1941_1950": 1945,
    "1951_1960": 1955,
    "1961_1970": 1965,
    "1971_1980": 1975,
    "1981_1990": 1985,
    "1991_2000": 1995,
    "2001_2010": 2005,
    "2011_2015": 2013
}

age_counts_for_math["dominant_age_midpoint"] = (
    age_counts_for_math["dominant_age_period_label"]
    .map(dominant_age_midpoint)
)

display(
    age_counts_for_math[
        ["dominant_age_period_label", "dominant_age_midpoint"]
    ].head(15)
)
```

```python
# Midpoint years for each age period
age_period_midpoints = {
    "x_bis_1900": 1890,
    "x1901_1910": 1905,
    "x1911_1920": 1915,
    "x1921_1930": 1925,
    "x1931_1940": 1935,
    "x1941_1950": 1945,
    "x1951_1960": 1955,
    "x1961_1970": 1965,
    "x1971_1980": 1975,
    "x1981_1990": 1985,
    "x1991_2000": 1995,
    "x2001_2010": 2005,
    "x2011_2015": 2013
}

# Weighted average construction year
weighted_age_sum = sum(
    age_counts_for_math[col] * midpoint
    for col, midpoint in age_period_midpoints.items()
)

age_counts_for_math["weighted_average_construction_year"] = np.where(
    age_counts_for_math[age_period_cols].sum(axis=1) > 0,
    weighted_age_sum / age_counts_for_math[age_period_cols].sum(axis=1),
    np.nan
)

display(
    age_counts_for_math[
        [
            "dominant_age_period_label",
            "dominant_age_midpoint",
            "weighted_average_construction_year"
        ]
    ].head(15)
)
```

```python
age_counts_for_math["weighted_average_age_2015"] = (
    2015 - age_counts_for_math["weighted_average_construction_year"]
)

age_counts_for_math["dominant_age_2015"] = (
    2015 - age_counts_for_math["dominant_age_midpoint"]
)

display(
    age_counts_for_math[
        [
            "dominant_age_period_label",
            "weighted_average_construction_year",
            "weighted_average_age_2015",
            "dominant_age_2015"
        ]
    ].head(15)
)
```

```python
# Earliest and latest represented construction periods

age_presence = age_counts_for_math[age_period_cols] > 0

earliest_period_col = age_presence.replace(
    False, np.nan
).idxmax(axis=1)

latest_period_col = age_presence.iloc[:, ::-1].replace(
    False, np.nan
).idxmax(axis=1)

age_counts_for_math["earliest_age_midpoint"] = (
    earliest_period_col.map(age_period_midpoints)
)

age_counts_for_math["latest_age_midpoint"] = (
    latest_period_col.map(age_period_midpoints)
)

# Set records with no age information to NaN
no_age_data = age_counts_for_math[age_period_cols].sum(axis=1) == 0

age_counts_for_math.loc[no_age_data, [
    "earliest_age_midpoint",
    "latest_age_midpoint"
]] = np.nan

# Temporal range of the represented distribution
age_counts_for_math["age_distribution_range"] = (
    age_counts_for_math["latest_age_midpoint"]
    - age_counts_for_math["earliest_age_midpoint"]
)

display(
    age_counts_for_math[
        [
            "earliest_age_midpoint",
            "latest_age_midpoint",
            "age_distribution_range"
        ]
    ].head(15)
)
```

```python
positive_cases = df["reporting_of_demolition"].eq("reported")

missingness_by_target = pd.DataFrame({
    "total_records": [
        len(df),
        positive_cases.sum(),
        (~positive_cases).sum()
    ],
    "missing_age": [
        df["number_of_age_periods"].eq(0).sum(),
        (positive_cases & df["number_of_age_periods"].eq(0)).sum(),
        ((~positive_cases) & df["number_of_age_periods"].eq(0)).sum()
    ],
    "missing_density_change": [
        df["pop_density_pressure_change"].isna().sum(),
        (positive_cases & df["pop_density_pressure_change"].isna()).sum(),
        ((~positive_cases) & df["pop_density_pressure_change"].isna()).sum()
    ],
    "complete_age_density": [
        (
            df["number_of_age_periods"].gt(0)
            & df["pop_density_pressure_change"].notna()
        ).sum(),
        (
            positive_cases
            & df["number_of_age_periods"].gt(0)
            & df["pop_density_pressure_change"].notna()
        ).sum(),
        (
            (~positive_cases)
            & df["number_of_age_periods"].gt(0)
            & df["pop_density_pressure_change"].notna()
        ).sum()
    ]
}, index=["all", "reported", "not_reported"])

display(missingness_by_target)
```

```python
from scipy.stats import pointbiserialr

continuous_predictors = [
    "bldg_area",
    "bldg_volume",
    "energy_intensity_kwh_m3a",
    "pop_density_pressure_change",
    "pop_density_pressure_pct_change",
    "dominant_age_share",
    "dominant_age_midpoint",
    "weighted_average_construction_year",
    "weighted_average_age_2015",
    "dominant_age_2015",
    "age_distribution_range",
    "age_data_available",
    "density_change_available",
]

corr_rows = []

for col in continuous_predictors:
    valid = df[[col, "demolition_target"]].dropna()

    if valid[col].nunique() > 1:
        r, p = pointbiserialr(
            valid["demolition_target"],
            valid[col]
        )
    else:
        r, p = np.nan, np.nan

    corr_rows.append({
        "variable": col,
        "n": len(valid),
        "positive_cases_in_n": int(valid["demolition_target"].sum()),
        "correlation": r,
        "p_value": p,
        "missing": int(df[col].isna().sum())
    })

correlation_results = (
    pd.DataFrame(corr_rows)
    .sort_values(
        "correlation",
        key=lambda s: s.abs(),
        ascending=False
    )
)

display(correlation_results)
```

```python
bty_demolition_summary = (
    df.groupby("BtySig")
      .agg(
          total_buildings=("building_serial_id", "count"),
          reported_demolitions=("demolition_target", "sum")
      )
      .reset_index()
)

bty_demolition_summary["reporting_rate_per_1000"] = (
    bty_demolition_summary["reported_demolitions"]
    / bty_demolition_summary["total_buildings"]
    * 1000
)

bty_demolition_summary["share_of_reported_demolitions_pct"] = (
    bty_demolition_summary["reported_demolitions"]
    / bty_demolition_summary["reported_demolitions"].sum()
    * 100
)

bty_demolition_summary = bty_demolition_summary.sort_values(
    "reporting_rate_per_1000",
    ascending=False
)

bty_demolition_summary
```

```python
btysig_labels = {
    "SFH": "Single-family houses",
    "MFH": "Multi-family houses",
    "HOT": "Hotels and restaurants",
    "HSN": "Hospitals",
    "ONR": "Other non-residential buildings",
    "OFC": "Offices",
    "TRD": "Trade buildings",
    "PRO": "Factories and workshops",
    "STG": "Storage buildings",
    "ASB": "Agricultural service buildings",
    "NAS": "Other non-agricultural service buildings"
}

bty_demolition_summary["building_type_name"] = (
    bty_demolition_summary["BtySig"].map(btysig_labels)
)

bty_demolition_summary = bty_demolition_summary[
    [
        "BtySig",
        "building_type_name",
        "total_buildings",
        "reported_demolitions",
        "reporting_rate_per_1000",
        "share_of_reported_demolitions_pct"
    ]
]

bty_demolition_summary
```

```python
import numpy as np
import pandas as pd

z = 1.96  # 95% confidence level

ci_results = []

for _, row in bty_demolition_summary.iterrows():
    x = int(row["reported_demolitions"])
    n = int(row["total_buildings"])

    if n == 0:
        lower = np.nan
        upper = np.nan
    else:
        p = x / n

        denominator = 1 + (z**2 / n)

        centre = (
            p + (z**2 / (2 * n))
        ) / denominator

        margin = (
            z * np.sqrt(
                (p * (1 - p) / n)
                + (z**2 / (4 * n**2))
            )
        ) / denominator

        lower = centre - margin
        upper = centre + margin

    ci_results.append({
        "BtySig": row["BtySig"],
        "building_type_name": row["building_type_name"],
        "total_buildings": n,
        "reported_demolitions": x,
        "rate_per_1000": row["reporting_rate_per_1000"],
        "lower_ci_per_1000": lower * 1000,
        "upper_ci_per_1000": upper * 1000
    })

bty_rate_ci = pd.DataFrame(ci_results)

bty_rate_ci = bty_rate_ci.sort_values(
    "rate_per_1000",
    ascending=False
)

bty_rate_ci
```
